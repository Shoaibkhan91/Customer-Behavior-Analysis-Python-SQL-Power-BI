```python
"""I use group-wise median imputation. I group the data by Category, calculate the median 
Review_Rating within each category, and use that category-specific median to fill missing values."""
```

```python
#customer["Review_Rating"] = customer.groupby("Category")["Review_Rating"].transform(lambda x: x.fillna(median()))
median_rating=customer.groupby("Category").Review_Rating.mean()
customer["Review_Rating"] = customer["Review_Rating"].fillna(
    customer["Category"].map(median_rating)
)
```

```python
# Removing Extra space in Columns
customer.columns=customer.columns.str.replace(" ","_")
customer.columns = customer.columns.str.lower()
 customer.rename(columns={"purchase_amount_(usd)":"purchase_amount"	},inplace=True )
```

```python
# creating a olumns ag _group 
labels =["Young_Adult" , "Adult" , "Middle_age" ,"Senior"]
customer["Age_group"] = pd.qcut(customer.age,q=4 ,labels=labels)
```

```python
# create columns purchase_frequency_days

frequency_mapping = {
    "Fortnightly":14,
    "Annually":365,
    "Weekly":7,
    "Bi-Weekly":14,
    "Monthly":30,
    "Quarterly":90,
    "Every 3 Months":90
}
customer["purchase_frequency_days"] = customer["frequency_of_purchases"].map(frequency_mapping)
```
